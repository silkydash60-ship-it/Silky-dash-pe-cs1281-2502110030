IT08

```python
logs = [
    "P101 25.5 accepted",
    "P102 26.2 rework",
    "P103 24.8 scrap",
    "P104 27.1 accepted",
    "P105 25.9 rework",
    "P106 26.5 accepted",
    "P107 24.6 scrap",
    "P108 27.3 accepted",
    "P109 25.2 rework",
    "P110 26.8 scrap"
]

accepted = []
rework = []
scrap = []
scrap_codes = []

for line in logs:
    data = line.split()
    code = data[0]
    dia = float(data[1])
    status = data[2]

    if status == "accepted":
        accepted.append(dia)
    elif status == "rework":
        rework.append(dia)
    elif status == "scrap":
        scrap.append(dia)
        scrap_codes.append(code)

print("Accepted count:", len(accepted))
print("Rework count:", len(rework))
print("Scrap count:", len(scrap))

print("Accepted average:", sum(accepted) / len(accepted))

print("Scrapped part codes:", scrap_codes)
```

```python
day_logs = [
    "P101 25.5 accepted",
    "P102 26.2 scrap",
    "P103 24.8 rework",
    "P104 27.1 accepted",
    "P105 25.9 scrap"
]

night_logs = [
    "P101 25.5 scrap",
    "P102 26.2 accepted",
    "P106 26.5 accepted",
    "P107 24.6 scrap",
    "P108 27.3 accepted"
]

day_scrap = set()
night_scrap = set()

day_accept = 0
night_accept = 0

for line in day_logs:
    data = line.split()
    code = data[0]
    status = data[2]

    if status == "accepted":
        day_accept += 1
    elif status == "scrap":
        day_scrap.add(code)

for line in night_logs:
    data = line.split()
    code = data[0]
    status = data[2]

    if status == "accepted":
        night_accept += 1
    elif status == "scrap":
        night_scrap.add(code)

print("Day Accept Rate:", day_accept / len(day_logs) * 100, "%")
print("Night Accept Rate:", night_accept / len(night_logs) * 100, "%")

print("Scrap in Both Shifts:", day_scrap & night_scrap)
print("Unique Day Scrap:", day_scrap - night_scrap)
print("Unique Night Scrap:", night_scrap - day_scrap)
print("All Unique Scrap Codes:", day_scrap ^ night_scrap)
```
