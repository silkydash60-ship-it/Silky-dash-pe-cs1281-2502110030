```python
class Machine:
    def __init__(self, name, capacity, rating):
        self.name = name
        self.capacity = capacity
        self.rating = rating

    def __str__(self):
        return self.name + " - Capacity: " + str(self.capacity)

    def run(self, jobs):
        return jobs <= self.capacity


m = Machine("Press-1", 200, 5)

print(m)
print("Rating:", m.rating)
print("Can process 150 jobs:", m.run(150))
```

```python
class Machine:
    def __init__(self, name):
        self.name = name
        self.__state = "STOPPED"

    def start(self):
        if self.__state == "STOPPED":
            self.__state = "RUNNING"
            print("Machine started")
        else:
            print("Machine is already running")

    def stop(self):
        if self.__state == "RUNNING":
            self.__state = "STOPPED"
            print("Machine stopped")
        else:
            print("Machine is already stopped")

    def status(self):
        return self.__state


m = Machine("Press-1")

print("Status:", m.status())
m.start()
print("Status:", m.status())
m.stop()
print("Status:", m.status())
```

```python
class Machine:
    def __init__(self, name):
        self.name = name
        self.parts = 0
        self.jobs = 0

    def process(self, job):
        self.jobs += 1
        self.parts += job
        print(self.name, "processed", job, "parts")

    def status(self):
        return self.parts


m = Machine("Press-1")

jobs = [10, 20, 15, 25]

for job in jobs:
    m.process(job)

print("\nShift Report")
print("Machine:", m.name)
print("Jobs:", m.jobs)
print("Total parts:", m.status())
```

```python
class Machine:
    def __init__(self, name):
        self.name = name
        self.parts = 0
        self.busy = False

    def run(self, job):
        self.busy = True
        self.parts += job
        print(self.name, "processed", job, "parts")
        self.busy = False


machines = [
    Machine("Press-1"),
    Machine("Press-2"),
    Machine("Press-3")
]

jobs = [20, 30, 15, 25, 10]

for job in jobs:
    for m in machines:
        if not m.busy:
            m.run(job)
            break

print("\nPlant Report")

total = 0

for m in machines:
    print(m.name, ":", m.parts, "parts")
    total += m.parts

print("Plant Total:", total)
```
