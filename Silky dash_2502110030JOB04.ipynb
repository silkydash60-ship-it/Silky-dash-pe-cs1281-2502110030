IT10

```python
import random

accept = 0
rework = 0
scrap = 0

for i in range(100):
    dia = round(random.uniform(24.90, 25.10), 2)

    if 24.95 <= dia <= 25.05:
        print(dia, "ACCEPT")
        accept += 1
    elif 24.90 <= dia < 24.95 or 25.05 < dia <= 25.10:
        print(dia, "REWORK")
        rework += 1
    else:
        print(dia, "SCRAP")
        scrap += 1

print("SHIFT REPORT")
print("Accepted:", accept)
print("Rework:", rework)
print("Scrap:", scrap)
print("Total:", accept + rework + scrap)

print("Same code, ten times the work, zero extra effort.")
```

```python
import random

mode = input("Enter mode (halt/skip): ").lower()

accept = 0
rework = 0
scrap = 0
inspected = 0

for i in range(100):
    dia = round(random.uniform(24.90, 25.10), 2)
    inspected += 1

    if 24.95 <= dia <= 25.05:
        print(dia, "ACCEPT")
        accept += 1

    elif dia < 24.95 or dia > 25.05:
        if dia == 24.90 or dia == 25.10:
            print(dia, "SCRAP")
            scrap += 1

            if mode == "halt":
                print("Batch stopped!")
                break
            elif mode == "skip":
                continue
        else:
            print(dia, "REWORK")
            rework += 1

print("\nSHIFT REPORT")
print("Inspected:", inspected)
print("Accepted:", accept)
print("Rework:", rework)
print("Scrap:", scrap)
```

```python
import random

consecutive = 0
attempts = 0
max_attempts = 500

while consecutive < 5 and attempts < max_attempts:
    dia = round(random.uniform(24.90, 25.10), 2)
    attempts += 1

    if 24.95 <= dia <= 25.05:
        print(dia, "ACCEPT")
        consecutive += 1
    else:
        print(dia, "REJECT")
        consecutive = 0

if consecutive == 5:
    print("Five consecutive ACCEPTS achieved!")
else:
    print("Safety limit reached!")

print("Total parts inspected:", attempts)
print("Safety cap:", max_attempts)
print("A safety cap prevents an infinite loop and protects the machine from running endlessly.")
```
