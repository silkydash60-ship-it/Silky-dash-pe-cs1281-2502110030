```python
def parse_data(data):
    rejects = []
    values = []

    for i, x in enumerate(data):
        try:
            n = float(x)

            if n < 0:
                raise ValueError("Negative value")

            values.append(n)

        except ValueError as e:
            rejects.append((i + 1, x, str(e)))

    if values:
        average = sum(values) / len(values)
    else:
        average = 0

    return values, rejects, average


data = ["10", "20", "abc", "-5", "30"]

values, rejects, average = parse_data(data)

print("Valid:", values)
print("Rejected:", rejects)
print("Average:", average)
```

```python
class SpecViolation(Exception):
    pass

class Machine:
    def __init__(self, minimum, maximum):
        self.minimum = minimum
        self.maximum = maximum

    def check_value(self, value):
        if value < self.minimum or value > self.maximum:
            raise SpecViolation(
                f"Value {value} is outside specification "
                f"({self.minimum} - {self.maximum})"
            )

        return "Value accepted"


machine = Machine(10, 50)

test_values = [20, 45, 60]

for value in test_values:
    try:
        print(value, ":", machine.check_value(value))
    except SpecViolation as error:
        print(value, ":", error)
```

```python
def check(x):
    try:
        n = float(x)

        if n >= 0:
            return True
        return False

    except ValueError:
        return False


def run_tests():
    tests = ["10", "-5", "abc", "20"]

    for x in tests:
        if check(x):
            print(x, "PASS")
        else:
            print(x, "FAIL")
run_tests()
```
