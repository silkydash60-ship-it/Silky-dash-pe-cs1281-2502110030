IT10

```python
accept = 0
rework = 0
scrap = 0

for i in range(10):
    d = float(input("Enter diameter of part: "))

    if 24.95 <= d <= 25.05:
        print("ACCEPT")
        accept += 1

    elif 24.90 <= d <= 25.10:
        print("REWORK")
        rework += 1

    else:
        print("SCRAP")
        scrap += 1

print("\n----- BATCH REPORT -----")
print("Accepted:", accept)
print("Rework:", rework)
print("Scrap:", scrap)
```

IT08

```python
total = int(input("Enter total number of parts: "))
size = int(input("Enter box size: "))

boxes = total // size
loose = total % size

print("Full boxes:", boxes)
print("Loose parts:", loose)
print("Audit:", boxes * size + loose == total)
```

```python
accept = 0
rework = 0
scrap = 0

for i in range(10):
    d = float(input("Enter diameter: "))

    if 24.95 <= d <= 25.05:
        print("ACCEPT")
        accept += 1
    elif 24.90 <= d <= 25.10:
        print("REWORK")
        rework += 1
    else:
        print("SCRAP")
        scrap += 1

boxes = accept // 4
loose = accept % 4

print("\n----- END OF SHIFT SUMMARY -----")
print("Accepted parts:", accept)
print("Rework parts:", rework)
print("Scrap parts:", scrap)
print("Full boxes:", boxes)
print("Loose parts:", loose)
print("Audit:", boxes * 4 + loose == accept)
```
