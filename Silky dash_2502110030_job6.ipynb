```python
def inspect(dia):
    if dia > 25.05:
        return "oversize"
    elif dia < 24.95:
        return "undersize"
    else:
        return "OK"

def tally(labels):
    counts = {"OK": 0, "undersize": 0, "oversize": 0}
    for x in labels:
        counts[x] += 1
    return counts

def report(shift, counts):
    print(shift, "accept", counts["OK"], "rework", counts["undersize"], "scrap", counts["oversize"])

data = [25.02, 24.90, 25.05]
labels = []

for x in data:
    labels.append(inspect(x))

counts = tally(labels)
report("DAY", counts)
```

```python
def inspect(dia):
    if dia > 25.05:
        return "oversize"
    elif dia < 24.95:
        return "undersize"
    return "OK"

def tally(labels):
    counts = {"OK": 0, "undersize": 0, "oversize": 0}
    for x in labels:
        counts[x] += 1
    return counts

def report(shift, counts):
    print(shift, "accept", counts["OK"], "rework", counts["undersize"], "scrap", counts["oversize"])

day = [25.02, 24.90, 25.05]
night = [25.09, 24.98, 24.80]

report("DAY", tally([inspect(x) for x in day]))
report("NIGHT", tally([inspect(x) for x in night]))
```

```python
def lookup(catalogue, code):
    return catalogue.get(code, 0)

def scrap_codes(catalogue):
    scrap = set()
    for code, dia in catalogue.items():
        if dia > 25.05 or dia < 24.95:
            scrap.add(code)
    return scrap

catalogue = {
    "P101": 25.02,
    "P102": 24.90,
    "P103": 25.09,
    "P104": 24.98
}

print(catalogue)
print(lookup(catalogue, "P101"))
print(lookup(catalogue, "P999"))
print(scrap_codes(catalogue))
```
