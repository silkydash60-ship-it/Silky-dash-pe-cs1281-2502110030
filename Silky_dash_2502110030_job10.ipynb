```python
import os

# Use relative paths
filename = "data.txt"

try:
    # Open the file using a relative path
    with open(filename, "r") as file:
        content = file.read()
        print("File contents:")
        print(content)

except FileNotFoundError:
    print("Error: The file was not found.")

except PermissionError:
    print("Error: Permission denied.")

except OSError as e:
    print(f"Error: Unable to access the file. {e}")

print("Program completed.")
```

```python
import os

# Create reports folder
os.makedirs("reports", exist_ok=True)

# Function to calculate results
def calculate_marks(marks):
    if marks < 0 or marks > 100:
        return "Invalid Input"
    elif marks >= 40:
        return "PASS"
    else:
        return "FAIL"

# Run tests
def run_tests():
    assert calculate_marks(80) == "PASS"
    assert calculate_marks(25) == "FAIL"
    assert calculate_marks(40) == "PASS"
    assert calculate_marks(-5) == "Invalid Input"
    assert calculate_marks(110) == "Invalid Input"
    print("All tests passed successfully!")

run_tests()

# Generate and save sample report
report = """
STUDENT RESULT REPORT
---------------------
Marks: 80
Result: PASS

Marks: 25
Result: FAIL
"""

report_path = os.path.join("reports", "result_report.txt")
with open(report_path, "w", encoding="utf-8") as f:
    f.write(report)

print("\nReport saved successfully to reports/result_report.txt! Details below:")
print(report)
```
