```python

import csv
import os

os.makedirs("data", exist_ok=True)

valid_records = []
rejected_records = []

with open("data/shift_log.txt", "r") as file:
    for line_no, line in enumerate(file, 1):
        try:
            parts = [x.strip() for x in line.strip().split(",")]

            if len(parts) != 4:
                raise ValueError("Invalid number of fields")

            wo, process, time, status = parts

            if not wo or not process or not time or not status:
                raise ValueError("Missing field")

            time = float(time)

            if time <= 0:
                raise ValueError("Invalid time")

            if status not in ["Completed", "Pending"]:
                raise ValueError("Invalid status")

            valid_records.append([wo, process, time, status])

        except ValueError as e:
            rejected_records.append([line_no, line.strip(), str(e)])

with open("data/report.txt", "w") as file:
    file.write(f"{'Work Order':<15}{'Process':<15}{'Time':<12}{'Status'}\n")
    file.write("-" * 55 + "\n")

    for wo, process, time, status in valid_records:
        file.write(f"{wo:<15}{process:<15}{time:<12}{status}\n")

with open("data/rejects.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Line No", "Rejected Record", "Reason"])
    writer.writerows(rejected_records)

print("Pipeline completed successfully!")
print("Valid Records:", len(valid_records))
print("Rejected Records:", len(rejected_records))
print("Report saved in data/report.txt")
print("Rejected records saved in data/rejects.csv")
```

```python

import re

with open("data/shift_log.txt", "r") as file:
    lines = file.readlines()

pattern = r"^\s*(WO\d{2})\s*\|\s*(.*?)\s*\|\s*(PASS|FAIL|NORMAL|COMPLETE)\s*$"

valid = []
rejected = []

for line in lines:


    match = re.match(pattern, line, re.IGNORECASE)

    if match:
        wo, task, status = match.groups()

        if task.strip():
            valid.append([wo, task.strip(), status.upper()])
        else:
            rejected.append(line.strip())
    else:
        rejected.append(line.strip())

print("VALID RECORDS:")
for record in valid:
    print(record)

print("\nREJECTED RECORDS:")
for record in rejected:
    print(record)

print("\nTotal lines:", len(lines))
print("Valid lines:", len(valid))
print("Rejected lines:", len(rejected))

print("\nRegex parsing survived more test lines because it")
print("handles spacing and validates the work order and status.")
```

```python
import os
import csv
from datetime import datetime

DATA_DIR = "data"
REPORT_DIR = "data/reports"

os.makedirs(REPORT_DIR, exist_ok=True)

timestamp = datetime.now().astimezone().isoformat(timespec="seconds")

total_processed = 0
total_rejected = 0

summary = []

for filename in sorted(os.listdir(DATA_DIR)):

    if not filename.endswith(".txt"):
        continue

    with open(f"{DATA_DIR}/{filename}", "r") as file:
        records = [line.strip() for line in file if line.strip()]

    valid = [r for r in records if len(r) >= 3]
    rejected = [r for r in records if len(r) < 3]

    processed_count = len(valid)
    rejected_count = len(rejected)

    total_processed += processed_count
    total_rejected += rejected_count

    with open(f"{REPORT_DIR}/{filename}_report.txt", "w") as report:
        report.write(f"File: {filename}\n")
        report.write(f"Timestamp: {timestamp}\n")
        report.write(f"Records Processed: {processed_count}\n")
        report.write(f"Rejects: {rejected_count}\n")
        report.write(f"Valid Records: {valid}\n")
        report.write(f"Rejected Records: {rejected}\n")

    with open("data/run_history.csv", "a", newline="") as history:
        writer = csv.writer(history)

        if history.tell() == 0:
            writer.writerow([
                "timestamp", "file_name",
                "records_processed", "rejects"
            ])

        writer.writerow([
            timestamp, filename,
            processed_count, rejected_count
        ])

    summary.append(
        f"{filename}: Processed={processed_count}, Rejected={rejected_count}"
    )

with open("data/plant_summary.txt", "w") as report:
    report.write("COMBINED PLANT SUMMARY\n")
    report.write(f"Timestamp: {timestamp}\n")
    report.write(f"Total Files: {len(summary)}\n")
    report.write(f"Total Processed: {total_processed}\n")
    report.write(f"Total Rejected: {total_rejected}\n\n")

    report.write("\n".join(summary))

print("BATCH JOB COMPLETED SUCCESSFULLY!")
print("Total Processed:", total_processed)
print("Total Rejected:", total_rejected)
print("Reports saved in:", REPORT_DIR)
print("Run history saved in: data/run_history.csv")
```
